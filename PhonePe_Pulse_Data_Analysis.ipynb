```python
import matplotlib.pyplot as plt

# Find latest year and quarter
latest_year = State_TxnSplit["Year"].max()
latest_quarter = State_TxnSplit[State_TxnSplit["Year"] == latest_year]["Quarter"].max()

# Filter most recent quarter data
recent_data = State_TxnSplit[
    (State_TxnSplit["Year"] == latest_year) &
    (State_TxnSplit["Quarter"] == latest_quarter)
]

# Distribution of transaction types per state
txn_distribution = recent_data.groupby(["State", "Transaction Type"])["Transactions"].sum().reset_index()

# Plot distribution for each state
for st in txn_distribution["State"].unique():
    state_txn = txn_distribution[txn_distribution["State"] == st]
    plt.bar(state_txn["Transaction Type"], state_txn["Transactions"])
    plt.title(f"Transaction Type Distribution - {st} (Year {latest_year}, Q{latest_quarter})")
    plt.xticks(rotation=45)
    plt.ylabel("Transactions")

    # Show exact numbers above each bar
    for x, y in zip(state_txn["Transaction Type"], state_txn["Transactions"]):
        plt.text(x, y + 1000, str(y), ha="center", va="bottom")

    plt.show()
```
