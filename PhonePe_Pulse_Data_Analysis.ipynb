```python
import numpy as np
import pandas as pd

# District totals
district_totals = District_Txn_and_Users.groupby("State")[["Transactions","Amount (INR)","Registered Users"]].sum().reset_index()

# State totals
state_totals = state_txn_and_users.groupby("State")[["Transactions","Amount (INR)","Registered Users"]].sum().reset_index()

# Merge datasets
df_compare = pd.merge(district_totals, state_totals, on="State", suffixes=("_dist", "_state"))

# Define match conditions (use correct column names with suffixes)
tx_match     = df_compare["Transactions_dist"] == df_compare["Transactions_state"]
amount_match = np.isclose(df_compare["Amount (INR)_dist"], df_compare["Amount (INR)_state"])
users_match  = df_compare["Registered Users_dist"] == df_compare["Registered Users_state"]

# Perfect matches = all three conditions true
perfect_match = tx_match & amount_match & users_match

# Discrepancies = rows that are NOT perfect matches
discrepancies = df_compare[~perfect_match]

# Display results
print("--- Discrepancies Found Using AND (&) Logic ---")
if discrepancies.empty:
    print("Success: All data points match perfectly!")
else:
    print(discrepancies[["State",
                         "Transactions_dist","Transactions_state",
                         "Amount (INR)_dist","Amount (INR)_state",
                         "Registered Users_dist","Registered Users_state"]])
```

```python
#Data Visualisation 
#5.1.1
import matplotlib.pyplot as plt

# 1. Select a state (example: Karnataka)
state_name = "Karnataka"
state_data = state_txn_and_users[state_txn_and_users["State"] == state_name]

# 2. Group by Year & Quarter
time_data = state_data.groupby(["Year","Quarter"]).agg({
    "Transactions":"sum",
    "Amount (INR)":"sum"
}).reset_index()

# 3. Create line chart with markers
plt.plot(time_data["Year"].astype(str) + "-Q" + time_data["Quarter"].astype(str),
         time_data["Transactions"],
         label="Transactions",
         marker="o", linestyle="-", color="blue")

plt.plot(time_data["Year"].astype(str) + "-Q" + time_data["Quarter"].astype(str),
         time_data["Amount (INR)"],
         label="Amount (INR)",
         marker="s", linestyle="--", color="green")

# 4. Formatting
plt.xticks(rotation=45)
plt.legend()
plt.title(f"Transactions count & Amount Over Time for - {state_name}")
plt.xlabel("Year-Quarter")
plt.ylabel("Value")
plt.show()
```
